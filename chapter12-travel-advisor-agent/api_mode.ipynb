```python
import requests

BASE_URL = "http://127.0.0.1:8000"
APP_NAME = "travel_advisor_agent"
USER_ID = "lucas_123"
SESSION_ID = ""

response = requests.post(f"{BASE_URL}/apps/{APP_NAME}/users/{USER_ID}/sessions")

data = response.json()

SESSION_ID = data.get('id')
print(SESSION_ID)
```

```python
message = {
    "appName": APP_NAME,
    "userId": USER_ID,
    "sessionId": SESSION_ID,
    "newMessage": {"parts": [{"text": "Yes, I want to know the currency exchange rate"}],"role": "user"},
}

response = requests.post(f"{BASE_URL}/run_sse", json=message)

data = response.json()

for event in data:
    content = event.get("content")
    parts = content.get("parts")
    for part in parts:
        function_call = part.get("functionCall", None)
        if function_call:
            print(function_call.get("name"))
        text = part.get("text", None)
        if text:
            print(text)
        print("|||||" * 30)
    print("=" * 60)
```

```python
import sseclient
import json

message = {
    "appName": APP_NAME,
    "userId": USER_ID,
    "sessionId": SESSION_ID,
    "newMessage": {"parts": [{"text": "What is weather there?"}],"role": "user"},
    "streaming": True
}

response = requests.post(f"{BASE_URL}/run_sse", json=message, stream=True)

client = sseclient.SSEClient(response)

for event in client.events():
    data = json.loads(event.data)
    content = data.get("content")
    parts = content.get("parts")
    for part in parts:
        function_call = part.get("functionCall", None)
        if function_call:
            print(function_call.get("name"))
        text = part.get("text", None)
        if text:
            print(text)
        print("|||||" * 60)
    print("=" * 60)
```
